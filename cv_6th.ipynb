```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Load the image in grayscale
img = cv2.imread('temple.jpeg', cv2.IMREAD_GRAYSCALE)

# Check if image loaded successfully
if img is None:
    print("Error: Image not found.")
else:

    # Define different averaging kernels
    kernel_3x3 = np.ones((3, 3), np.float32) / 9
    kernel_5x5 = np.ones((5, 5), np.float32) / 25
    kernel_7x7 = np.ones((7, 7), np.float32) / 49

    # Apply averaging filters
    blur_3x3 = cv2.filter2D(img, -1, kernel_3x3)
    blur_5x5 = cv2.filter2D(img, -1, kernel_5x5)
    blur_7x7 = cv2.filter2D(img, -1, kernel_7x7)

    # Display results
    plt.figure(figsize=(12, 8))

    # Original Image
    plt.subplot(2, 2, 1)
    plt.imshow(img, cmap='gray')
    plt.title("Original Image")
    plt.axis('off')

    # 3x3 Filter
    plt.subplot(2, 2, 2)
    plt.imshow(blur_3x3, cmap='gray')
    plt.title("3x3 Averaging Filter")
    plt.axis('off')

    # 5x5 Filter
    plt.subplot(2, 2, 3)
    plt.imshow(blur_5x5, cmap='gray')
    plt.title("5x5 Averaging Filter")
    plt.axis('off')

    # 7x7 Filter
    plt.subplot(2, 2, 4)
    plt.imshow(blur_7x7, cmap='gray')
    plt.title("7x7 Averaging Filter")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
